#### 🐍 Conceito de Python · As regras do Python: o que pode e o que não pode

O Python é muito exigente. Ele não "adivinha" o que você quis dizer.

| ✅ Pode | ❌ Não pode | Por quê |
|---|---|---|
| `preco_arroz = 25.90` | `preco arroz = 25.90` | Nome de variável **não tem espaço**. Use `_` para separar palavras. |
| `item1 = 5` | `1item = 5` | Nome de variável **não começa com número**. |
| `preco = 25.90` | `preço = 25.90` *(evite)* | Acento até funciona, mas dá erro fácil de digitação. Prefira sem acento. |
| `print("Olá")` | `Print("Olá")` | **Maiúscula e minúscula são diferentes.** `print` ≠ `Print`. |
| `print("Olá")` | `print("Olá)` | Toda **aspa** e todo **parêntese** que abre tem que fechar. |
| `25.90` | `25,90` | Decimal é com **ponto**. |
| `1500` | `1.500` | Milhar **sem ponto**. `1.500` vira "um vírgula cinco". |
| linha começando encostada à esquerda | espaços sobrando no começo da linha | Espaço no começo da linha tem significado no Python. Só use quando o professor pedir. |

**Quando aparecer um erro**, leia a **última linha** da mensagem vermelha. Os mais comuns:

| Erro | O que quer dizer | Como resolver |
|---|---|---|
| `NameError` | O Python não conhece esse nome | Confira se digitou igual (maiúsculas também) ou se executou a célula onde ele foi criado |
| `SyntaxError` | Algo escrito de um jeito que o Python não entende | Procure aspa ou parêntese sem fechar, ou vírgula no lugar de ponto |
| `IndentationError` | Espaço sobrando ou faltando no começo da linha | Ajuste os espaços no começo da linha |

✏️ **Tente você:** na célula do passo 4, troque `print` por `Print` e execute. Leia o erro. Depois desfaça a troca e execute de novo.

```python
import matplotlib.pyplot as plt

def grafico_ia(ia, x, y, novo, nome_x, nome_y, titulo):
    valores_x = [v[0] for v in x]
    eixo = [[i] for i in range(int(min(valores_x) * 0.7), int(max(valores_x) * 1.3) + 1)]
    previsao = ia.predict([[novo]])[0]
    plt.figure(figsize=(8, 4.5))
    plt.___(valores_x, y, s=90, color="#164194", label="Exemplos (o passado)", zorder=3)  # ← lacuna 1
    plt.plot([e[0] for e in eixo], ia.predict(eixo), color="#164194", alpha=0.35, linewidth=3, label="Regra que a IA aprendeu")
    plt.scatter([novo], [previsao], s=240, color="#E84910", marker="*", label=f"Previsão para {novo}: {previsao:.1f}", zorder=4)
    plt.xlabel(nome_x)
    plt.ylabel(nome_y)
    plt.title(titulo)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

grafico_ia(ia_sorvete, temperatura, sorvetes, ___, "Temperatura (°C)", "Sorvetes vendidos", "Mercadinho: IA analítica prevê as vendas de sorvete")  # ← lacuna 2
```
