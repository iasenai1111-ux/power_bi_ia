### Passo 8 · Ver a IA prever o próximo token

**🎯 O que vamos fazer:** descobrir o segredo da IA generativa: ela escreve **um token por vez**, sempre escolhendo o próximo mais provável.

**✍️ Complete as lacunas `___`** na célula logo abaixo, junto com o professor. Apague o `___` e escreva no lugar:

1. **Lacuna 1** → **quantos** tokens mostrar no gráfico: **`8`**
2. **Lacuna 2** → o **começo da frase**: **`"A obra atrasou por causa da"`**

**🔎 Passo a passo:**

| Trecho | O que ele faz |
|---|---|
| `import torch` | **PyTorch**: a biblioteca que faz as contas da IA por dentro. |
| `def proximo_token(texto):` | Uma função que mostra o que a IA "pensa" em escrever depois de um texto. |
| `modelo(**entrada)` | Manda o texto para a IA e recebe uma nota para **cada token que ela conhece** (são mais de 150 mil). |
| `torch.softmax(...)` | Transforma essas notas em **chances**, em porcentagem. |
| `torch.topk(prob, quantos)` | Pega só os 8 tokens com mais chance. |
| `plt.barh(...)` | Desenha um gráfico de barras deitadas. Você conheceu o matplotlib no Exercício 01. |

**👀 O que vai aparecer:** um gráfico de barras com os 8 tokens mais prováveis para continuar *"A obra atrasou por causa da"* e a chance de cada um.

```python
import torch
import matplotlib.pyplot as plt

def proximo_token(texto, quantos=___):  # ← lacuna 1
    modelo = ia_generativa.model
    entrada = tokenizador(texto, return_tensors="pt").to(modelo.device)
    with torch.no_grad():
        logits = modelo(**entrada).logits[0, -1].float()
    prob = torch.softmax(logits, dim=-1)
    top = torch.topk(prob, quantos)
    rotulos = [f'"{tokenizador.decode([int(i)])}"' for i in top.indices]
    valores = (top.values * 100).tolist()
    plt.figure(figsize=(8, 4))
    plt.barh(rotulos[::-1], valores[::-1], color="#164194")
    plt.xlabel("Chance (%)")
    plt.title(f"Próximo token depois de: {texto}")
    plt.grid(axis="x", alpha=0.3)
    plt.show()

proximo_token(___)  # ← lacuna 2
```

#### 🐍 Conceito de Python · for e if: repetir e decidir

Para conferir a resposta da IA, vamos usar duas ferramentas novas:

| Comando | O que faz | No dia a dia |
|---|---|---|
| `for` | **Repete** um passo para cada item de uma lista. | Passar cada produto do carrinho no caixa. |
| `if` | **Decide**: se a condição for verdade, faz uma coisa. | Se o produto custa mais de R$ 8, é caro. |
| `else` | O contrário do `if`: "se não for verdade, faz outra coisa". | Senão, é barato. |

Igual à função, `for`, `if` e `else` terminam com **dois-pontos** e as linhas de dentro levam **4 espaços** no começo.
